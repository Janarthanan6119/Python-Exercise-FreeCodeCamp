Exercise 4

Make a function that receives a list as parameter and return how many elements it has; if it doesn't have any element return an error.

```python
def lis(params):
    n = len(params)
    if(n==0):
        return 'An Error'
    else:
        return 'It has {} elements'.format(n)
```

```python
lis([7,2,5,9])
```

```python
lis([])
```
